# 체크포인트 생성 테스트

`checkpoints/pretrain_800m` (step 90522) 모델을 불러와서 직접 프롬프트를 입력해보고 생성 결과를 확인한다.

- 모델/토크나이저는 맨 위 셀에서 한 번만 로드한다 (느림, 재실행 불필요).
- 아래 "프롬프트 입력" 셀만 원하는 만큼 반복해서 값 바꿔가며 실행하면 됨.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import torch
import sentencepiece as spm

from src.eval.generate_sample import load_model, generate, DEFAULT_TOKENIZER, DEFAULT_MODEL_CONFIG

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)
```

```python
CHECKPOINT_DIR = ROOT / "checkpoints/pretrain_800m"

sp = spm.SentencePieceProcessor(model_file=str(ROOT / DEFAULT_TOKENIZER))
model = load_model(CHECKPOINT_DIR, ROOT / DEFAULT_MODEL_CONFIG, device)
print("모델/토크나이저 로드 완료")
```

## 프롬프트 입력

`prompt`, `max_new_tokens`, `temperature`, `top_p` 값만 바꿔서 이 셀만 반복 실행.

```python
prompt = "대한민국의 수도는"
max_new_tokens = 100
temperature = 0.8
top_p = 0.9

text = generate(model, sp, prompt, device,
                max_new_tokens=max_new_tokens, temperature=temperature, top_p=top_p)
print(text)
```

## 여러 프롬프트 한 번에 비교

```python
prompts = [
    "김해시 장유는",
]

for p in prompts:
    out = generate(model, sp, p, device, max_new_tokens=80, temperature=0.8, top_p=0.9)
    print(f"[prompt] {p}")
    print(f"[output] {out}")
    print("-" * 40)
```
