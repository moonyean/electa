```python
from pathlib import Path
import sentencepiece as spm

# 현재 폴더와 상위 폴더를 순회하며 프로젝트 루트를 찾는다.
roots = [Path.cwd(), *Path.cwd().parents]

# 프로젝트 위치를 직접 지정한 경로도 후보에 추가한다.
roots.append(Path(r"C:\Users\User\Desktop\trans"))

model_path = None

for root in roots:
    candidate = root / "data" / "processed" / "tokenizer" / "unigram_32k.model"
    if candidate.exists():
        model_path = candidate
        break

if model_path is None:
    raise FileNotFoundError(
        "unigram_32k.model 파일을 찾을 수 없습니다. "
        "프로젝트의 data/processed/tokenizer 폴더를 확인하세요."
    )

print("토크나이저 경로:", model_path)

# SentencePiece Unigram 모델 불러오기
tokenizer = spm.SentencePieceProcessor(
    model_file=str(model_path)
)

text = "안녕하세요. 한국어 토크나이저 테스트입니다."

# 토큰 문자열
tokens = tokenizer.encode(text, out_type=str)

# 토큰 ID
token_ids = tokenizer.encode(text, out_type=int)

# 다시 문장으로 복원
decoded_text = tokenizer.decode(token_ids)

print("원문:", text)
print("토큰:", tokens)
print("토큰 ID:", token_ids)
print("복원:", decoded_text)
print("어휘 크기:", tokenizer.vocab_size())
```
